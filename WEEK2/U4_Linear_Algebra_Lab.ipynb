```python
import numpy as np

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]
for i, (u, v) in enumerate(pairs, start=1):
    l1_norm = np.linalg.norm(u, ord=1)
    l2_norm = np.linalg.norm(u, ord=2)
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    cosine_sim = dot_product / (norm_u * norm_v)
    print(f"Pair {i}:")
    print(f"  u - L1 Norm: {l1_norm:.2f}, L2 Norm: {l2_norm:.2f}")
    print(f"  Cosine Similarity: {cosine_sim:.4f}\n")
    #most familiar pair is pair 1
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])
M_inv = np.linalg.inv(M)
M_trace = np.trace(M)
print("Inverse of M:\n", M_inv)
print(f"Trace of M: {M_trace}")
identity_check = M @ M_inv
is_identity = np.allclose(identity_check, np.eye(2))
print(f"\nM @ inv(M) matches the Identity Matrix: {is_identity}")
is_M_symmetric = np.allclose(M, M.T)
is_P_orthogonal = np.allclose(P.T @ P, np.eye(2))
print(f"\nIs M symmetric? {is_M_symmetric}")
print(f"Is P orthogonal? {is_P_orthogonal}")
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])
S = np.array([[2.0, 0.0],
              [0.0, 0.5]])
theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
tri_closed = np.hstack([tri, tri[:, [0]]])
tri_transformed = R @ S @ tri
tri_trans_closed = np.hstack([tri_transformed, tri_transformed[:, [0]]])
plt.figure(figsize=(6, 6))
plt.plot(tri_closed[0, :], tri_closed[1, :], 'b-o', label='Original')
plt.plot(tri_trans_closed[0, :], tri_trans_closed[1, :], 'r-o', label='Transformed (Scaled + Rotated)')
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.grid(True, linestyle=':', alpha=0.6)
plt.axis('equal')
plt.legend()
plt.title('2D Geometric Transformation')
plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np

C = np.array([[4., 1.],
              [2., 3.]])
eigenvalues, eigenvectors = np.linalg.eig(C)
print("Eigenvalues:", eigenvalues)
lam1 = eigenvalues[0]
v1 = eigenvectors[:, 0]

Av = C @ v1
lam_v = lam1 * v1

is_verified = np.allclose(Av, lam_v)
print(f"\nVerification (A @ v == lambda * v): {is_verified}")
print("A @ v:     ", Av)
print("lambda * v:", lam_v)
```

```python
import numpy as np
A2 = np.array([[3., 2.],
               [1., 4.]])
b2 = np.array([7., 9.])
cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])
x = np.linalg.solve(A2, b2)
print("Solution x:", x)
rank = np.linalg.matrix_rank(A2)
print("Rank of A2:", rank)
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

cos_cat_dog = cosine_similarity(cat, dog)
cos_cat_car = cosine_similarity(cat, car)

print(f"\nCosine similarity (cat, dog): {cos_cat_dog:.4f}")
print(f"Cosine similarity (cat, car): {cos_cat_car:.4f}")

if cos_cat_dog > cos_cat_car:
    print("Conclusion: 'cat' and 'dog' are more similar.")
else:
    print("Conclusion: 'cat' and 'car' are more similar.")
```
