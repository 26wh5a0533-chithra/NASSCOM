```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)

```

```python
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
import sympy as sp

def g(x):
    return x**3 + 2*x
x_val = 2.0
h = 1e-6
numerical_deriv = (g(x_val + h) - g(x_val - h)) / (2 * h)
print(f"Numerical Derivative (at x=2): {numerical_deriv:.6f}")
x = sp.Symbol('x')
g_sym = x**3 + 2*x
symbolic_deriv_expr = sp.diff(g_sym, x)
print(f"Symbolic Derivative Expression: {symbolic_deriv_expr}")
symbolic_deriv_val = float(symbolic_deriv_expr.subs(x, 2))
print(f"Symbolic Derivative Value (at x=2): {symbolic_deriv_val:.6f}")
difference = abs(numerical_deriv - symbolic_deriv_val)
print(f"Absolute difference: {difference:.2e}")
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
h2 = x**2 * y + sp.sin(y)
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print(f"dh/dx: {dh_dx}")
print(f"dh/dy: {dh_dy}")
gradient = [dh_dx, dh_dy]
print(f"Gradient vector expression: {gradient}")
point = {x: 2, y: 0}
evaluated_gradient = [float(component.subs(point)) for component in gradient]
print(f"Evaluated gradient at (2, 0): {evaluated_gradient}")
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
import sympy as sp
x = sp.Symbol('x')
y = sp.exp(x**2 + 1)
inner = x**2 + 1
inner_prime = 2 * x
outer_prime = sp.exp(inner)
by_hand_deriv = outer_prime * inner_prime
print(f"Manual Chain Rule expression: {by_hand_deriv}")
auto_deriv = sp.diff(y, x)
print(f"Automated sp.diff expression: {auto_deriv}")
are_matching = sp.simplify(by_hand_deriv - auto_deriv) == 0
print(f"\nDo the manual and automated derivatives match? {are_matching}")
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np
np.random.seed(42)
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1
Wb = np.random.randn(8, 1) * 0.1
def relu(x): return np.maximum(0, x)
def relu_derivative(x): return (x > 0).astype(float)
z1 = Xb @ Wa
h = relu(z1)
y_hat = h @ Wb
loss_before = np.mean((y_hat - Yb) ** 2)
m = Xb.shape[0]
dy_hat = 2 * (y_hat - Yb) / m
dWb = h.T @ dy_hat
dh = dy_hat @ Wb.T
dz1 = dh * relu_derivative(z1)
dWa = Xb.T @ dz1
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb
z1_new = Xb @ Wa
h_new = relu(z1_new)
y_hat_new = h_new @ Wb
loss_after = np.mean((y_hat_new - Yb) ** 2)

print(f"Loss before optimization step: {loss_before:.6f}")
print(f"Loss after optimization step:  {loss_after:.6f}")
```

```python
xv = 0.0
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```
